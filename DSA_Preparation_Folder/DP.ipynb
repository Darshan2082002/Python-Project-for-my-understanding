### Example of Dynamic Programming

```python
def feb(n):
    if n<=1:
        return n
    return feb(n-1)+feb(n-2)

if __name__=="__main__":
    n=40

    print(feb(n))
```

Memoization Technique


```python
def fibrec(n,memo):
    if n<=1:
        return n 
    if memo[n]!=-1:
        return memo[n]
    memo[n]=fibrec(n-1,memo)+fibrec(n-2,memo)

    return memo[n]

def fib(n):
    memo=[-1]*(n+1)
    return fibrec(n,memo)

if __name__=="__main__":
    n=40
    print(fib(n))
```

Tabulization method 


```python
def fib(n):
    dp=[0]*(n+1)
    dp[0]=0
    dp[1]=1

    for i in range(2,n+1):
        dp[i]=dp[i-1]+dp[i-2]
    return dp[n]

if __name__=="__main__":
    n=40
    print(fib(n))
```

```python
def DNA_1(n,m):

    dp=[]
    for _ in range(m):
        row=[0]*(n)
        dp.append(row)
    return dp

print(DNA_1(3,3))
```

```python
def DNA_2(n,m):
    dp=[[0]*(n) for _ in range(m)] 
    return dp 
print(DNA_2(3,3))
```

```python
def DNA():
    n=input().strip()
    m=input().strip()
    l1=len(n)
    l2=len(m)
    dp=[[0]*(l2+1) for _ in range(l1+1)]
    for i in range(1,l1+1):
        for j in range(1,l2+1):
            if n[i-1]==m[j-1]:
                dp[i][j]=1+dp[i-1][j-1]
            else:
                dp[i][j]=max(dp[i-1][j],dp[i][j-1])
    return(dp[l1][l2])
print(DNA())
```
